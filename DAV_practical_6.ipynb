## **Practical 6|Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).**

**Step1: import library**

```python
#Import library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Create new dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West","North", "South", "East", "West"],
    "Sales": [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold": [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
```

```python
#Covert data into dataframe
df = pd.DataFrame(data)
#Display dataframe
df
```

**Step: 2 Create a Line Chart**
q: how did the companys sales change from jan to dec

```python
#Set graph size
plt.figure(figsize=(8, 4))
#Calculate monthly sales
Monthly_sales = df.groupby('Month')['Sales'].sum()
#Plot month and sales
plt.plot(Monthly_sales.index, Monthly_sales.values, marker='o')
#Give name to X-axis
plt.xlabel('Month')
#Give name to Y-axis
plt.ylabel('Sales')
#Give title
plt.title('Company Sales from January to December')
#Show grid
plt.grid(True)
#Display graph
plt.show()
```

**Step: 3 Create a Bar Chart**
q: which product generate how much total sales

```python
#Set graph size
plt.figure(figsize=(8, 4))
#Calculate total sales for each product
Product_sales = df.groupby('Product')['Sales'].sum()
#Plot product and total sales
plt.bar(Product_sales.index, Product_sales.values)
#Give name to X-axis
plt.xlabel('Product')
#Give name to Y-axis
plt.ylabel('Total Sales')
#Give title
plt.title('Total Sales by Product')
#Display graph
plt.show()
```

**Step: 4 Create a Histogram**
q: how are the individual sales values distributed

```python
#Set graph size
plt.figure(figsize=(8, 4))
#Plot individual sales values
plt.hist(df['Sales'], bins=10)
#Give name to X-axis
plt.xlabel('Sales')
#Give name to Y-axis
plt.ylabel('Frequency')
#Give title
plt.title('Distribution of Individual Sales Values')
#Display graph
plt.show()
```

**Step: 5 Create a Scatter Plot**
q: is there a relationship bet the no of units sold and the total sales

```python
#Set graph size
plt.figure(figsize=(8, 4))
#Plot units sold and total sales
plt.scatter(df['Units_Sold'], df['Sales'])
#Give name to X-axis
plt.xlabel('Number of Units Sold')
#Give name to Y-axis
plt.ylabel('Total Sales')
#Give title
plt.title('Units Sold vs Total Sales')
#Show graph
plt.grid(True)
#Display graph
plt.show()
```

**Step: 6 Create a Pie Chart**
q: what percentage of total sales comes from each region

```python
plt.figure(figsize=(8, 4))
#Calculate total sales for each region
Region_sales = df.groupby('Region')['Sales'].sum()
#Plot pie chart
plt.pie(Region_sales.values, labels=Region_sales.index, autopct='%1.1f%%')
#Give title
plt.title('Percentage of Total Sales by Region')
#Show graph
plt.show()
```
