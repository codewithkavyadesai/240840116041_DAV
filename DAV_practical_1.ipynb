**Practical 1 | Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON ƒles using Pandas.**

**Step:1 Import Pandas Liberary**

```python
# Step 1: Import the Pandas library
import pandas as pd
```

**Step:2 create simple data**

```python
# Store student details such as ID, Name, Age, City and Marks
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krishna", "Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gandhinagar", "Ahmedabad"],
    "Marks": [85, 92, 78, 88, 81, 84, 90]
}
```

**Step:3 create dataframe**

```python
# DataFrame stores the data in rows and columns
df = pd.DataFrame(data)
df
```

**Step:4 save data as csv file **

```python
# index=False prevents saving the row numbers
df.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

**Step:5 save data as excel file **

```python
# Creates an Excel file containing the student data
df.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")
```

**Step:6 save data as json file**

```python
# orient="records" stores each row as a separate record
# indent=4 makes the JSON file easy to read
df.to_json("students.json", orient="records", indent=4)
print("JSON file created successfully.")
```

**Step:7 read the csv file**
