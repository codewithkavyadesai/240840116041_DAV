## Practical-3 | Apply fltering, sorting, aggregation, and GroupBy operations on real-world datasets.

**Step:1 Import library**

```python
# Import library
import pandas as pd
```

```python
# Create employee dataset
data ={
    "Emp_Id":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Convert data into DataFrame
df = pd.DataFrame(data)
df
```

**Step:2 Filtter Data**

```python
# Filter employees from IT department
df[df['Department']=="IT"]
```

```python
# Filter employees from Surat city
df[df['City']=="Surat"]
```

```python
# Filter IT employees having salary more than 50000
df[(df["Department"]=="IT")&(df["Salary"]>50000)]
#Both conditions must be True:
#1. Department must be "IT"
#2. Salary must be greater than 50,000
```

**Step:3 Sort Data**

```python
# Sort data according to salary
df.sort_values(by="Salary")
```

```python
# Sort salary from highest to lowest
df.sort_values(by="Salary",ascending=False)
```

```python
# Sort data by department and salary
df.sort_values(by=["Department","Salary"], ascending=[True,False])
```

**Step:4 Perfrom Aggregation**

```python
# Find total salary
df["Salary"].sum()
```

```python
# Find average salary
df["Salary"].mean()
```

```python
# Find highest salary
df["Salary"].max()
```

```python
# Find lowest salary
df["Salary"].min()
```

```python
# Count total employees
df["Emp_Id"].count()
```

**Step:5 Perfrom Group By Operation**

```python
# Find average salary of each department
df.groupby("Department")["Salary"].mean()
```

```python
# Find minimum salary of each department
df.groupby("Department")["Salary"].min()
```

```python
# Count employees in each department
df.groupby("Department")["Emp_Id"].count()
```

```python
# Find maximum salary of each department
df.groupby("Department")["Salary"].max()
```

**Step:6 Multiple Aggerration Function**

```python
# Find count, total, average, minimum and maximum salary
# for each department
df.groupby("Department")["Salary"].agg(["count","sum","mean","min","max"])
```

**Step:7 Save Your Result**

```python
# Save the employee data into CSV file
df.to_csv("Employee_Data.csv",index=False)
# Show message after creating the dataset
print("Dataset create successfully.")
```
