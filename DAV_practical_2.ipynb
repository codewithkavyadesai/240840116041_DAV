### **Practical-2 | Perfrom data inspection, cleaning, handling missing values duplicate removal, and data transformation using pandas.**

**Step:1 Import Pandas library**

```python
# Import the Pandas library
import pandas as pd
```

**Step:2 Create Simple Dataset**

```python
# Create a dictionary containing student data
data = {
    "ID":[101, 102, 103, 104, 104, 105, 106, 107],
    "Name" :["Rahu", "priya", "Amit", None, "Riya", "Ajay", " Naresh", "Jenil"],
    "Age" :[20, 21, None, 20, 20, 22, 23, 28],
    "City" :["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahmedabad"],
    "Marks" : [85, None, 78, 88, 88, 81, 90, 93]
}
```

```python
# Create a DataFrame from the data
df = pd.DataFrame(data)
# Display dataframe
df
```

**Step:3 Inspect the dataset**

```python
# Display the first five rows of the dataset.
df.head()
```

```python
# Display the last five rows of the dataset.
df.tail()
```

```python
# Display detailed information about the dataset.
df.info()
```

```python
# Display Statistical Information for numerical columns.
df.describe()
```

```python
# Display the dimensions of the dataset.
df.shape
```

```python
#Display the data types of each columns.
df.dtypes
```

```python
# Display all the column of the dataframe.
df.columns
```

**Step:4 Check Missing Values**

```python
# Check for missing (null) values in every column.
df.isnull()
```

```python
# count null values for each column.
df.isnull().sum()
```

```python
# count total number of null values.
df.isnull().sum().sum()
```

**Step:5 Fill Missing Values**

```python
#replace missing values in the marks column.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

```python
# replace missing values in the age column.
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
# replacing missing name with the text "unknown".
df["Name"] = df["Name"].fillna("unknown")
df
```

**STep:6 Setect Duplicate records**

```python
# Check whether each raw is duplicated.
df.duplicated()
```

```python
# count the total number of duplicate rows.
df.duplicated().sum()
```

**Step:7 Remove Duplicate Records**

```python
# Remove duplicate rows from the DataFrame
df.drop_duplicates()
df
```

**Step:8 Rename a column**

```python
# rename the column "Marks" to "score".
df.rename(columns={"Marks":"Score"}, inplace=True)
# Display the DataFrame
df
```

**Step:9 change Data Type**

```python
# cover the Age column from float  to integer.
df.astype({"Age":"int64"}).dtypes
```

**Step:10 Create a New column**

```python
# Create a new column called "Final Score"
# Add 5 marks to the Score column
df["Final Score"]=df["Score"]+5
df
```

**Step:11 Convert Text To UpperCase**

```python
# Convert all names to uppercase
df["Name"]=df["Name"].str.upper()
df
```

**Step:12 Covert Text To LowerCase**

```python
# Convert all city names to lowercase
df["City"]=df["City"].str.lower()
df
```

**Step:13 Remove Extra Spaces**

```python
# Remove extra spaces from the City column
df["City"] = df["City"].str.strip()
df
```

**Step:14 sort the dataset**

```python
# Sort the DataFrame based on the "Final Score" column
# ascending=False sorts the values from highest to lowest
df.sort_values(by="Final Score", ascending=False)
df
```

**Step:15 Save the cleaned dataset**

```python
# Save the cleaned df as a new csv file
df.to_csv("cleaned_student.csv", index=False)

# Display a success message
print("Dataset saved Successfully.")
```
