## **Practical 8| Create interactive visualizations using Plotly Express and Plotly Graph Objects.**

**Task:1 Import Library**

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

**Task:2 Create the Common Dataset**

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West','North', 'South', 'East', 'West'],
    'Sales': [120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    'Units_Sold': [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40],
    'Profit': [18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
}

df = pd.DataFrame(data)
df
```

```python
#Display first five records
df.head()
```

```python
#Display the number of rows and columns
df.shape
```

```python
#Display column names and datatpe
df.info()
```

**Task:3 Crete an Intractive Line Chart Using Plotly Express**

```python
#How does sales revenue change throughout the year?
fig = px.line(df, x='Month', y='Sales', title='Sales Revenue Over Time')
fig.show()
```

**Task:4 Improve the Hover Information**

```python
#Create a line chart using Sales over Month
fig = px.line(df, x='Month', y='Sales', title='Sales Revenue Over Time',
              hover_data=['Units_Sold', 'Profit'],
              color_discrete_sequence=['blue'])
fig.show()
```

```python
#Add additional columns to the hover information. Students can see Product,Region, Units Sold and profit when they move the mouse over a data point.
fig = px.line(
    df,
    x="Month",
    y="Sales",
    color="Product",
    markers=True,
    hover_data=["Product", "Region", "Units_Sold", "Profit"]
)

fig.update_layout(
    title="Sales Over Time",
    xaxis_title="Month",
    yaxis_title="Sales"
)
```

**Task:5 Create an Intractive Bar chart**

```python
#Group the dataset according to product using bar chart
product_data = df.groupby("Product", as_index=False)["Sales"].sum()
# Create bar chart
fig = px.bar(
    product_data,
    x="Product",
    y="Sales",
    title="Total Sales by Product",
    labels={
        "Product": "Product",
        "Sales": "Total Sales"
    },
    text_auto=True
)
fig.show()
```

```python
#Calculate the total sales generated by each Product
product_data = df.groupby("Product", as_index=False)["Sales"].sum()

# Create bar chart
fig = px.bar(
    product_data,
    x="Product",
    y="Sales",
    title="Sales by Product"
)

fig.show()
```

```python
#Create an intractive bar chart
fig = px.bar(
    product_data,
    x="Product",
    y="Sales",
    title="Interactive Sales by Product",
    text_auto=True,
    hover_data=["Product", "Sales"]
)

fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales"
)

fig.show()
```

**Task:6 Intractive Scatter Plot**

```python
#To analyze the relationship between Units Sold and sales using an interactive Sactter Plot
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    hover_data=["Product", "Region", "Units_Sold", "Sales", "Profit"],
    title="Relationship Between Units Sold and Sales",
    labels={
        "Units_Sold": "Units Sold",
        "Sales": "Sales"
    }
)

fig.update_traces(marker=dict(size=10))

fig.show()
```

**Task:7 Intractive Scatter Plot with color**

```python
#we will use region as the color variable
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Region",
    hover_data=["Product", "Region", "Units_Sold", "Sales", "Profit"],
    title="Relationship Between Units Sold and Sales",
    labels={
        "Units_Sold": "Units Sold",
        "Sales": "Sales"
    }
)

fig.update_traces(marker=dict(size=10))

fig.show()
```

**Task:8 Create an intractive pie chart**

```python
#To Visualize regional sales contribution using an intractive pie chart.
fig = px.pie(
    df,
    names="Region",
    values="Sales",
    title="Regional Sales Contribution",
    labels={
        "Region": "Region",
        "Sales": "Total Sales"
    }
)

fig.show()
```

**Task:9 Plotly Graph Objects**

```python
#To Create an intractive visualization using plotly graph objects and understand how it differs from plotly Express.
# Create a figure object
fig = go.Figure()

# Add a line trace to the figure
fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Sales"],
        mode="lines",
        name="Sales"
    )
)

# Customize the layout
fig.update_layout(
    title="Interactive Sales Visualization",
    xaxis_title="Month",
    yaxis_title="Sales"
)

# Show the figure
fig.show()
```

**Task:11 Add multiple Traces**

```python
#Create an empty figure
fig = go.Figure()
#Add sales as the first trace
fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Sales"],
        mode="lines",
        name="Sales"
    )
)
#Add profit as the second trace
fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Profit"],
        mode="lines",
        name="Profit"
    )
)
#Customize the complete layout
fig.update_layout(
    title="Interactive Sales and Profit Visualization",
    xaxis_title="Month",
    yaxis_title="Amount"
)
#Show the figure
fig.show()
```

**Task 12: Create an Interactive Bar Chart Using Plotly Graph Objects**

```python
# Group data according to Product
product_data = df.groupby("Product", as_index=False)["Sales"].sum()

# Create an empty figure
fig = go.Figure()

# Add bar trace
fig.add_trace(
    go.Bar(
        x=product_data["Product"],
        y=product_data["Sales"],
        name="Sales",
        text=product_data["Sales"],
        textposition="outside",
        hovertemplate="Product: %{x}<br>Sales: %{y:,.0f}<extra></extra>"
    )
)

# Customize layout
fig.update_layout(
    title="Total Sales by Product Using Graph Objects",
    xaxis_title="Product",
    yaxis_title="Total Sales"
)

fig.show()
```

**Task 13: Create an Interactive Pie Chart Using Graph **

```python
# Group data according to Region
region_data = df.groupby("Region", as_index=False)["Sales"].sum()

# Create pie chart
fig = go.Figure(
    data=[
        go.Pie(
            labels=region_data["Region"],
            values=region_data["Sales"],
            hole=0.3,
            hovertemplate="Region: %{label}<br>Sales: %{value:,.0f}<br>Share: %{percent}<extra></extra>"
        )
    ]
)

fig.update_layout(
    title="Regional Sales Contribution Using Graph Objects"
)

fig.show()
```

**Task 14: Add Interactive Dropdown Menu Using Graph Objects**

```python
# Create an empty figure
fig = go.Figure()

# Add Sales trace
fig.add_trace(
    go.Scatter(
        x=df["Month"],
        y=df["Sales"],
        mode="lines+markers",
        name="Sales"
    )
)

# Add dropdown menu
fig.update_layout(
    title="Monthly Sales",
    xaxis_title="Month",
    yaxis_title="Sales",
    updatemenus=[
        dict(
            buttons=[
                dict(
                    label="Sales",
                    method="update",
                    args=[
                        {"y": [df["Sales"]]},
                        {"title": "Monthly Sales", "yaxis": {"title": "Sales"}}
                    ]
                ),
                dict(
                    label="Profit",
                    method="update",
                    args=[
                        {"y": [df["Profit"]]},
                        {"title": "Monthly Profit", "yaxis": {"title": "Profit"}}
                    ]
                )
            ],
            direction="down",
            x=1,
            y=1.15
        )
    ]
)

fig.show()
```
